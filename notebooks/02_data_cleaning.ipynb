```python
from pathlib import Path
import sys
import pandas as pd
from IPython.display import display, Image, Markdown
ROOT = Path.cwd()
if not (ROOT / 'src').exists():
    ROOT = ROOT.parent
assert (ROOT / 'src').is_dir(), 'Start from the repository or notebooks folder'
sys.path.insert(0, str(ROOT / 'src'))
from cleaning import clean_data
from kpi import summarize, grouped, pareto, operation_capacity
raw = pd.read_csv(ROOT / 'data/raw/manufacturing_data.csv')
clean = clean_data(raw)
```

```python
display(clean[['actual_quantity','good_quantity','scrap_quantity','runtime_min']].describe())
assert not clean.duplicated(['date','shift','machine_id']).any()
assert (clean.actual_quantity == clean.good_quantity + clean.scrap_quantity).all()
print('All generated records satisfy the count and time balances.')
# Demonstrate rejection rather than silent removal.
try:
    clean_data(pd.concat([raw, raw.iloc[:1]]))
except ValueError as error:
    print('Expected validation error:', error)
```
